```python
!pip -q install pandas scikit-learn

import pandas as pd
from google.colab import files
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

# Upload the training file
print("Select optdigits.tra")
train_upload = files.upload()
train_file = next(
    (name for name in train_upload if name.lower().endswith(".tra")),
    None
)

# Upload the test file
print("Select optdigits.tes")
test_upload = files.upload()
test_file = next(
    (name for name in test_upload if name.lower().endswith(".tes")),
    None
)

if train_file is None:
    raise ValueError("The first upload must be the optdigits.tra training file.")

if test_file is None:
    raise ValueError("The second upload must be the optdigits.tes test file.")

def load_dataset(filename):
    # Each row contains 64 pixel values, then the digit label.
    data = pd.read_csv(filename, header=None)

    if data.shape[1] != 65:
        raise ValueError(
            f"{filename} should have 65 columns "
            f"(64 features and 1 label), but has {data.shape[1]}."
        )

    features = data.iloc[:, :64]
    labels = data.iloc[:, 64]
    return features, labels

# Load the datasets
x_train, y_train = load_dataset(train_file)
x_test, y_test = load_dataset(test_file)

# Train the SVM model
model = make_pipeline(
    StandardScaler(),
    SVC(kernel="rbf", C=10, gamma="scale")
)
model.fit(x_train, y_train)

# Predict the digits in the test dataset
predictions = model.predict(x_test)

# Show the results
print(f"Training examples: {len(x_train)}")
print(f"Test examples: {len(x_test)}")
print(f"Test accuracy: {accuracy_score(y_test, predictions):.2%}")

print("\nPer-digit results:")
print(classification_report(y_test, predictions, digits=3, zero_division=0))

print("Confusion matrix (rows = actual digit, columns = predicted digit):")
print(confusion_matrix(y_test, predictions, labels=list(range(10))))
```
